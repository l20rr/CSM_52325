```python
MSE = np.mean((Ia - Ib)**2)
print(MSE)
Ic = Ia * 1.
Id = Ib * 1.
MSE = np.mean((Ic.astype(np.float64) - Id.astype(np.float64))**2)
print(MSE)
```

```python
Prog = np.mean(Ic**2)
Perr = np.mean((Ic - Id)**2)
PSNR = 10 * np.log10(Prog / Perr)
print('PSNR (dBs):', PSNR)
```
